## 2. Environment Setup & Data Ingestion
We load our modular ingestion and statistical analysis components from `src.data_loader` and `src.analysis`.

```python
import sys
from pathlib import Path
import pandas as pd

# Ensure project root is in sys.path whether executed from root or inside notebooks/
project_root = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.data_loader import load_poll_data
from src.analysis import (
    get_dataset_overview,
    get_missing_value_summary,
    get_numeric_summary,
    get_categorical_frequency,
    get_department_language_crosstab,
    get_department_internship_crosstab,
    get_satisfaction_distribution,
    get_learning_method_distribution,
    get_recommendation_distribution,
    get_internship_interest_summary,
)

data_path = project_root / "data" / "processed" / "clean_poll_data.csv"
if not data_path.exists():
    raise FileNotFoundError(f"Expected cleaned dataset not found at: '{data_path}'")

df = load_poll_data(data_path)
print(f"Successfully loaded cleaned dataset: {len(df)} rows, {len(df.columns)} columns.")
```

## 6. Categorical Frequency Distributions
Analyzing marginal response proportions across department, programming language, learning method, and internship interest.

```python
print("--- Department Distribution ---")
display(get_categorical_frequency(df, "department"))

print("\n--- Preferred Language Distribution ---")
display(get_categorical_frequency(df, "preferred_language"))

print("\n--- Preferred Learning Method Distribution ---")
display(get_learning_method_distribution(df))

print("\n--- Internship Interest Distribution ---")
display(get_internship_interest_summary(df))
```

**Distribution Observations:**
- **Department Representation**: IT is the largest cohort (177 responses, 35.4%), followed by Commerce (131, 26.2%), Science (123, 24.6%), and Arts (69, 13.8%).
- **Programming Languages**: Python is the most frequently selected language (206, 41.2%), followed by Java (129, 25.8%), JavaScript (87, 17.4%), and C++ (78, 15.6%).
- **Internship Interest**: 372 respondents (74.4%) indicated interest in internships ('Yes'), while 128 (25.6%) responded 'No'.

## 7. Demographic Cross-Tabulations
Examining bivariate relationships between academic department and student preferences.

```python
print("--- Department vs. Preferred Language (Counts) ---")
display(get_department_language_crosstab(df))

print("\n--- Department vs. Preferred Language (Row Percentages %) ---")
display(get_department_language_crosstab(df, normalize="index"))

print("\n--- Department vs. Internship Interest (Counts) ---")
display(get_department_internship_crosstab(df))
```

**Cross-Tabulation Observations:**
- Across all four departments, Python consistently represents the most preferred programming language (IT: 39.5%, Commerce: 41.2%, Science: 40.7%, Arts: 46.4%).
- Internship interest remains high across all disciplines, ranging between 70.7% in Science to 78.0% in IT.

> **Caution:** These patterns are purely descriptive characteristics of the synthetic generation parameters and do not establish causal relationships.
